```python

import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Create a sample user-movie rating dataset
ratings = {
    'Movie A': [5, 4, 0, 0, 1],
    'Movie B': [4, 5, 0, 0, 2],
    'Movie C': [0, 0, 5, 4, 0],
    'Movie D': [0, 0, 4, 5, 0],
    'Movie E': [1, 2, 0, 0, 5]
}

df = pd.DataFrame(
    ratings,
    index=['User1', 'User2', 'User3', 'User4', 'User5']
)

# Calculate similarity between users
similarity = cosine_similarity(df)
similarity_df = pd.DataFrame(
    similarity,
    index=df.index,
    columns=df.index
)

# Select the target user
target_user = 'User1'

# Find similar users
similar_users = similarity_df[target_user].drop(target_user)

# Predict ratings using weighted average
predicted_ratings = {}

for movie in df.columns:
    if df.loc[target_user, movie] == 0:
        numerator = 0
        denominator = 0

        for user, score in similar_users.items():
            rating = df.loc[user, movie]
            if rating > 0:
                numerator += score * rating
                denominator += score

        if denominator > 0:
            predicted_ratings[movie] = numerator / denominator

# Recommend the highest-rated unseen movie
recommendations = sorted(
    predicted_ratings.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Recommendations for", target_user)

for movie, rating in recommendations:
    print(movie, "Predicted Rating:", round(rating, 2))
```
