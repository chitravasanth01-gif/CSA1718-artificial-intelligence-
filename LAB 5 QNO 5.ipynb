```python

import random

# Training sequence of weather events
events = [
    "Sunny", "Rainy", "Sunny", "Rainy",
    "Sunny", "Cloudy", "Sunny", "Rainy",
    "Sunny", "Cloudy"
]

# Get all unique states
states = list(set(events))

# Create transition count dictionary
transitions = {
    state: {next_state: 0 for next_state in states}
    for state in states
}

# Count transitions between events
for i in range(len(events) - 1):
    current = events[i]
    next_event = events[i + 1]
    transitions[current][next_event] += 1

# Calculate transition probabilities
probabilities = {}

for state in states:
    total = sum(transitions[state].values())
    probabilities[state] = {}

    for next_state in states:
        if total > 0:
            probabilities[state][next_state] = (
                transitions[state][next_state] / total
            )

# Display transition probabilities
print("Transition Probabilities:")

for state in states:
    print(state, ":", probabilities[state])

# Predict the next event
current_event = "Sunny"

next_events = probabilities[current_event]
predicted_event = max(next_events, key=next_events.get)

print("\nCurrent Event:", current_event)
print("Predicted Next Event:", predicted_event)
print("Prediction Probability:",
      round(next_events[predicted_event], 2))
```
