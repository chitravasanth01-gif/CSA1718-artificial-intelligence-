```python

import numpy as np
import random

# XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([[0], [1], [1], [0]])

# Neural network: 2 input, 4 hidden, 1 output
input_size = 2
hidden_size = 4

# Total number of weights and biases
gene_count = (
    input_size * hidden_size +
    hidden_size +
    hidden_size +
    1
)

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

# Neural network prediction
def predict(genes, inputs):
    i = 0

    W1 = genes[i:i+8].reshape(2, 4)
    i += 8

    b1 = genes[i:i+4]
    i += 4

    W2 = genes[i:i+4].reshape(4, 1)
    i += 4

    b2 = genes[i]

    hidden = np.tanh(np.dot(inputs, W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)

    return output

# Fitness function
def fitness(genes):
    predictions = predict(genes, X)
    error = np.mean((y - predictions) ** 2)
    return 1 / (error + 0.001)

# Create initial population
population_size = 100
population = [
    np.random.uniform(-2, 2, gene_count)
    for _ in range(population_size)
]

# Genetic Algorithm
for generation in range(1000):

    # Calculate fitness
    scores = [fitness(ind) for ind in population]

    # Sort population by fitness
    ranked = sorted(
        zip(scores, population),
        key=lambda item: item[0],
        reverse=True
    )

    # Keep the best two individuals
    new_population = [
        ranked[0][1].copy(),
        ranked[1][1].copy()
    ]

    # Generate new individuals
    while len(new_population) < population_size:

        # Selection
        p1 = random.choice(ranked[:20])[1]
        p2 = random.choice(ranked[:20])[1]

        # Crossover
        mask = np.random.rand(gene_count) < 0.5
        child = np.where(mask, p1, p2)

        # Mutation
        for i in range(gene_count):
            if random.random() < 0.1:
                child[i] += np.random.normal(0, 0.5)

        new_population.append(child)

    population = new_population

# Select the best neural network
best = max(population, key=fitness)

# Display final predictions
print("Genetic Algorithm Training Completed!")
print("\nXOR Predictions:")

predictions = predict(best, X)

for inputs, output in zip(X, predictions):
    result = 1 if output[0] >= 0.5 else 0
    print(inputs, "=", result,
          "Confidence:", round(output[0], 3))
```
