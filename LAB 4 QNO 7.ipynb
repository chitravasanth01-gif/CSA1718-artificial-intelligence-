```python

import random

# Fitness function
def fitness(x):
    return x ** 2

# Generate initial population
population = [random.randint(0, 31) for _ in range(6)]

for generation in range(20):

    # Selection: choose the best individuals
    population.sort(key=fitness, reverse=True)
    parents = population[:3]

    # Crossover: create new children
    children = []
    while len(children) < 3:
        p1, p2 = random.sample(parents, 2)
        child = (p1 + p2) // 2
        children.append(child)

    # Mutation
    for i in range(len(children)):
        if random.random() < 0.2:
            children[i] = random.randint(0, 31)

    # Create new population
    population = parents + children

    best = max(population, key=fitness)
    print("Generation:", generation + 1,
          "Best x:", best,
          "Fitness:", fitness(best))

print("\nOptimal Solution:", best)
print("Maximum Fitness:", fitness(best))
```
