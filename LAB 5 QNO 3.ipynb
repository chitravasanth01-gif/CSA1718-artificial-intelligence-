```python

import numpy as np
import random

# Environment settings
size = 5
states = size * size
actions = [0, 1, 2, 3]  # Up, Down, Left, Right

# Q-table
Q = np.zeros((states, len(actions)))

# Parameters
learning_rate = 0.8
discount = 0.95
epsilon = 0.2
episodes = 1000

goal = 24

# Function to get the next state
def move(state, action):
    row, col = divmod(state, size)

    if action == 0 and row > 0:
        row -= 1
    elif action == 1 and row < size - 1:
        row += 1
    elif action == 2 and col > 0:
        col -= 1
    elif action == 3 and col < size - 1:
        col += 1

    return row * size + col

# Train the agent
for episode in range(episodes):
    state = 0

    for step in range(100):
        # Choose an action
        if random.random() < epsilon:
            action = random.choice(actions)
        else:
            action = np.argmax(Q[state])

        # Perform the action
        next_state = move(state, action)

        # Calculate reward
        reward = 100 if next_state == goal else -1

        # Update Q-value
        Q[state, action] += learning_rate * (
            reward + discount * np.max(Q[next_state])
            - Q[state, action]
        )

        state = next_state

        if state == goal:
            break

# Test the trained agent
state = 0
path = [state]

for step in range(25):
    action = np.argmax(Q[state])
    state = move(state, action)
    path.append(state)

    if state == goal:
        break

print("Training Completed!")
print("Agent Path:", path)
print("Goal Reached:", state == goal)
```
