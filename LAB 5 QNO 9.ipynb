```python
!pip install deepface

import os
from deepface import DeepFace

# Load two input face images
image1 = "person1.jpg"
image2 = "person2.jpg"

# Check if files exist to avoid processing exceptions
if not os.path.exists(image1) or not os.path.exists(image2):
    print("Error: One or both of the target images are missing!")
    print(f"Please upload '{image1}' and '{image2}' to your Colab workspace files before running this cell.")
else:
    try:
        # Perform face recognition
        result = DeepFace.verify(
            img1_path=image1,
            img2_path=image2,
            model_name="Facenet",
            detector_backend="opencv"
        )

        # Display the result
        print("\n=== Face Recognition Result ===")

        if result["verified"]:
            print("Both images belong to the same person.")
        else:
            print("The images belong to different persons.")

        print("Distance:", round(result["distance"], 4))
        print("Similarity Threshold:", result["threshold"])

    except Exception as e:
        print("An error occurred during verification:", e)
```
